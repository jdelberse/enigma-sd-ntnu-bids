```python
import os
import re
import json
import pandas as pd
```

```python
root = os.getcwd().replace("/code/wrangling","")
subs = pd.read_csv(root+"/participants.tsv",sep="\t").participant_id
```

```python
for sub in subs:
    for ses in ["ses-b9","ses-sd9"]:
        for direct in ["dir-AP","dir-PA"]:
            path = f"{root}/{sub}/{ses}/fmap/{sub}_{ses}_acq-dwi_{direct}_epi.json"

            if os.path.exists(path) == False:
                continue

            with open(path, "r") as file:
                data = json.load(file)

            entry = data["IntendedFor"][0]
            new_entry = re.sub("ses-T[0-9]", ses, entry)
            data["IntendedFor"] = [new_entry]

            with open(path, "w") as file:
                json.dump(data, file, indent=4)
```
